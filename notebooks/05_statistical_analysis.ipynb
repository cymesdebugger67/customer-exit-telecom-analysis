## Interpretation

Among tested categorical features, the **derived `protection_level`** has the highest observed bias-corrected Cramér's V (approximately **0.412**). Among original source fields, **`Contract`** has one of the strongest associations (approximately **0.410**). **`tenure`** has the largest absolute rank-biserial effect among tested numeric fields (approximately **0.480**).
