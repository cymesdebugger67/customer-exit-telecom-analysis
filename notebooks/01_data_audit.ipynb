## Audit interpretation

The source has **7,043 customers and 21 original fields**, with no duplicate customer IDs or duplicate rows. The **11 blank `TotalCharges` values** are spaces in the raw text, so a simple null count does not detect them; they become missing after numeric conversion. All 11 records have zero tenure. The cleaning step **retains those customers** and leaves their total charges missing rather than assuming zero.
