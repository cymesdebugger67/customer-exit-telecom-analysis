## Interpretation

Observed churn is **42.71% for month-to-month contracts** and **45.29% for electronic-check payments**. Compare each rate with its **customer count, churned count, and share of all churn**: a high rate in a small group is not necessarily the largest concentration of exits. The support/security profile is a derived grouping, so its pattern should not be interpreted as an independent causal effect.
